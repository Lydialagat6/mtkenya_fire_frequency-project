```r
# load
library(readr)
library(dplyr)
library(tidyr)

```

```r
#read the data
library(readr)
temperature <- read.csv("raw_data/temp.csv")
head(temperature)
```

```r
#data cleaning

# Project: mtkenya_fire_frequency
# Purpose: to inspect the data and combine

# Inspect the dataset
View(temperature)
glimpse(temperature)
summary(temperature)


```

```r
# clean data
# Check duplicates
temp_clean <- temperature %>%
  mutate(
    year = as.integer(year),
    month = as.integer(month),
    temp_C_monthly_avg = as.numeric(temp_C_monthly_avg)
  ) %>%
  select(year, month, date, temp_C_monthly_avg) %>%
  arrange(date)
temp_clean
```

```r
# check missing values
colSums(is.na(temp_clean))
```

```r
# check duplicates
temp_clean %>%
  count(year, month) %>%
  filter(n > 1)
```

```r
# pivot longer
temp_wider <- temp_clean %>%
  pivot_wider(
    names_from = month,
    values_from = temp_C_monthly_avg)
temp_wide
head(temperature)
```
